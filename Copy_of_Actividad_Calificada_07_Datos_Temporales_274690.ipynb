## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# 1. Tipo de dato ANTES de convertir: pandas las lee como texto (object/str),
# porque el CSV no distingue fechas de cualquier otra cadena de caracteres.
print("Tipo de dato ANTES de convertir:")
print("pickup :", taxis['pickup'].dtype)
print("dropoff:", taxis['dropoff'].dtype)

# Contamos los faltantes antes de convertir, para poder comparar despues.
faltantes_antes = taxis['pickup'].isna().sum() + taxis['dropoff'].isna().sum()

# 2. Convertimos ambas columnas indicando el formato EXPLICITO que corresponde a los valores
# observados (por ejemplo '2019-03-23 20:21:09' -> '%Y-%m-%d %H:%M:%S'). Dar el formato
# explicito evita que pandas tenga que "adivinar" el formato fila por fila, lo cual es mas
# rapido y, sobre todo, mas seguro: si un valor no calza con el formato esperado, pandas
# lanzara un error en vez de interpretarlo de forma incorrecta y silenciosa.
taxis['pickup'] = pd.to_datetime(taxis['pickup'], format='%Y-%m-%d %H:%M:%S')
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format='%Y-%m-%d %H:%M:%S')

print("\nTipo de dato DESPUES de convertir:")
print("pickup :", taxis['pickup'].dtype)
print("dropoff:", taxis['dropoff'].dtype)

# 3. Verificamos que la conversion no haya generado NaT (fechas invalidas). Si el numero de
# faltantes es el mismo antes y despues, la conversion fue exitosa para todas las filas.
faltantes_despues = taxis['pickup'].isna().sum() + taxis['dropoff'].isna().sum()
print(f"\nFaltantes antes de convertir: {faltantes_antes}")
print(f"Faltantes (NaT) despues de convertir: {faltantes_despues}")
print("La conversion no genero NaT nuevos" if faltantes_antes == faltantes_despues else "¡Atencion! Se generaron NaT")
```

### 1.2 Periodo de los datos (5 pts)

El dataset debería contener solo viajes de marzo de 2019.

3. ¿Hay viajes con `pickup` en marzo y `dropoff` fuera de marzo? Cuéntalos.

```python
# 1. Fecha minima y maxima de pickup y dropoff.
print("pickup  -> minimo:", taxis['pickup'].min(), " maximo:", taxis['pickup'].max())
print("dropoff -> minimo:", taxis['dropoff'].min(), " maximo:", taxis['dropoff'].max())

# 2. Viajes cuyo pickup NO esta en marzo de 2019: usamos .dt.year y .dt.month para construir
# una mascara booleana y la negamos con "~" para quedarnos con lo que NO cumple la condicion.
pickup_en_marzo_2019 = (taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3)
fuera_de_marzo = ~pickup_en_marzo_2019
print(f"\nViajes con pickup fuera de marzo de 2019: {fuera_de_marzo.sum()}")
print(taxis.loc[fuera_de_marzo, ['pickup', 'dropoff']])

# 3. Viajes con pickup EN marzo pero dropoff FUERA de marzo (p. ej. un viaje que empieza el
# 31 de marzo tarde en la noche y termina ya en abril).
dropoff_fuera_de_marzo = ~((taxis['dropoff'].dt.year == 2019) & (taxis['dropoff'].dt.month == 3))
pickup_marzo_dropoff_fuera = pickup_en_marzo_2019 & dropoff_fuera_de_marzo
print(f"\nViajes con pickup en marzo pero dropoff fuera de marzo: {pickup_marzo_dropoff_fuera.sum()}")
print(taxis.loc[pickup_marzo_dropoff_fuera, ['pickup', 'dropoff']])
```

El 10 de marzo de 2019, en Nueva York, las horas entre 02:00 y 02:59 **no existieron**.

2. Localiza `pickup` en la zona `'America/New_York'` con `tz_localize()` usando `nonexistent='NaT'` y cuenta cuántos valores quedan como `NaT`.

```python
# 1. Filtramos los viajes cuyo pickup ocurrio el 10 de marzo de 2019 y contamos cuantos
# hay en cada hora del dia usando el accesor .dt.hour junto con value_counts().
viajes_10_marzo = taxis[taxis['pickup'].dt.date == pd.Timestamp('2019-03-10').date()]
conteo_por_hora = viajes_10_marzo['pickup'].dt.hour.value_counts().sort_index()

print(f"Total de viajes el 10 de marzo de 2019: {len(viajes_10_marzo)}")
print("\nViajes por hora ese dia:")
print(conteo_por_hora)
print("\n¿Hay viajes a las 2:00?", 2 in conteo_por_hora.index)

# 2. Localizamos pickup en la zona horaria 'America/New_York'. El parametro nonexistent='NaT'
# le dice a pandas que, si una marca de tiempo cae dentro del hueco que crea el cambio de
# horario (horas que nunca existieron, como las 02:xx del 10 de marzo de 2019), la convierta
# en NaT en lugar de intentar adivinar a que hora real corresponde.
pickup_localizado = taxis['pickup'].dt.tz_localize('America/New_York', nonexistent='NaT')
nat_generados = pickup_localizado.isna().sum()
print(f"\nValores NaT generados al localizar en America/New_York: {nat_generados}")
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# 1. Duracion como diferencia de Timestamps: pandas devuelve un Timedelta.
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# 2. Convertimos el Timedelta a minutos: dt.total_seconds() devuelve la duracion en segundos
# como numero flotante, y dividimos entre 60 para obtener minutos (permite valores fraccionarios,
# a diferencia de quedarnos solo con dt.seconds que trunca).
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

# 3. Resumen estadistico de la duracion en minutos.
print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).
- Duración **mayor a 3 horas**.

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
# Duracion negativa: el viaje "termina" antes de iniciar (dropoff < pickup), lo cual es
# fisicamente imposible y siempre se considera un error de registro.
duracion_negativa = (taxis['duracion_min'] < 0).sum()

# Duracion cero: pickup y dropoff exactamente iguales.
duracion_cero = (taxis['duracion_min'] == 0).sum()

# Duracion menor a 1 minuto: incluye los de duracion cero mas los que duraron entre 0 y 60 segundos.
duracion_menor_1min = (taxis['duracion_min'] < 1).sum()

# Duracion mayor a 3 horas (180 minutos): viajes anormalmente largos para un taxi urbano.
duracion_mayor_3h = (taxis['duracion_min'] > 180).sum()

print(f"Duracion negativa: {duracion_negativa}")
print(f"Duracion cero: {duracion_cero}")
print(f"Duracion menor a 1 minuto (incluye cero): {duracion_menor_1min}")
print(f"Duracion mayor a 3 horas: {duracion_mayor_3h}")

print("\nViajes con duracion cero:")
print(taxis.loc[taxis['duracion_min'] == 0, ['pickup', 'dropoff', 'distance', 'fare', 'total']])
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Calculamos la velocidad para TODOS los viajes primero (division vectorizada).
velocidad_para_todos = taxis['distance'] / (taxis['duracion_min'] / 60)

# 1. Con .where(condicion), conservamos el valor calculado solo donde la condicion es True
# (duracion_min > 0) y ponemos NaN donde es False (duracion 0, division indefinida/infinita).
taxis['velocidad_mph'] = velocidad_para_todos.where(taxis['duracion_min'] > 0)
print(f"Viajes con velocidad_mph faltante (duracion cero): {taxis['velocidad_mph'].isna().sum()}")

# 2. Viajes con velocidad mayor a 60 mph: en trafico urbano de Nueva York, mantener un promedio
# de mas de 60 mph durante todo un viaje es muy poco plausible y suele indicar un error en la
# distancia registrada o una duracion registrada demasiado corta.
viajes_muy_rapidos = (taxis['velocidad_mph'] > 60).sum()
print(f"Viajes con velocidad promedio mayor a 60 mph: {viajes_muy_rapidos}")

# 3. Los 5 viajes con mayor velocidad promedio, ordenados de mayor a menor.
columnas_vista = ['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']
top5_velocidad = taxis.sort_values('velocidad_mph', ascending=False).head(5)
print("\nTop 5 viajes con mayor velocidad promedio:")
print(top5_velocidad[columnas_vista])
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# hora: hora del dia en que inicio el viaje (0-23).
taxis['hora'] = taxis['pickup'].dt.hour

# dia_num: dia de la semana como numero, con lunes = 0 (comportamiento por defecto de dayofweek).
taxis['dia_num'] = taxis['pickup'].dt.dayofweek

# dia_nombre: nombre del dia de la semana en texto (en ingles, por defecto del locale de pandas).
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()

# fin_semana: True si el dia de la semana es sabado (5) o domingo (6).
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6])

# fecha: solo la parte de fecha de pickup, sin hora (util para contar dias distintos en 3.3).
taxis['fecha'] = taxis['pickup'].dt.date

# cruza_medianoche: True si la fecha de dropoff es distinta de la fecha de pickup, es decir,
# el viaje comenzo un dia y termino en un dia calendario diferente.
taxis['cruza_medianoche'] = taxis['dropoff'].dt.date != taxis['pickup'].dt.date

print(f"Viajes que cruzan la medianoche: {taxis['cruza_medianoche'].sum()}")
taxis[['pickup', 'dropoff', 'hora', 'dia_num', 'dia_nombre', 'fin_semana', 'fecha', 'cruza_medianoche']].head()
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Agrupamos por hora y calculamos, para cada una, el numero de viajes (size) y la mediana
# de la duracion en minutos. Usamos la mediana (no la media) porque, como se vio en 2.1 y 2.2,
# la duracion tiene valores extremos (viajes de 0 minutos y otros de mas de 100 minutos) que
# distorsionarian el promedio; la mediana es mas robusta a esos valores atipicos.
patron_por_hora = taxis.groupby('hora').agg(
    n_viajes=('hora', 'size'),
    mediana_duracion_min=('duracion_min', 'median')
)

print(patron_por_hora)

hora_mas_viajes = patron_por_hora['n_viajes'].idxmax()
hora_menos_viajes = patron_por_hora['n_viajes'].idxmin()
print(f"\nHora con MAS viajes: {hora_mas_viajes}:00 ({patron_por_hora['n_viajes'].max()} viajes)")
print(f"Hora con MENOS viajes: {hora_menos_viajes}:00 ({patron_por_hora['n_viajes'].min()} viajes)")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Agrupamos por dia_num y dia_nombre juntos: al incluir dia_num (numerico) como primer
# nivel de agrupacion, groupby ordena el resultado de lunes (0) a domingo (6) de forma natural.
patron_por_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    n_viajes=('dia_num', 'size'),
    mediana_duracion_min=('duracion_min', 'median'),
    mediana_fare=('fare', 'median')
).reset_index()

# Marzo de 2019 no tiene el mismo numero de lunes, martes, etc. (algunos dias de la semana
# aparecen 4 veces en el mes y otros 5), por lo que comparar solo el TOTAL de viajes por dia
# de la semana favoreceria a los dias que se repiten mas veces en el mes. Para corregir esto,
# contamos cuantas fechas distintas corresponden a cada dia de la semana con nunique()...
fechas_distintas_por_dia = taxis.groupby(['dia_num', 'dia_nombre'])['fecha'].nunique().reset_index(name='n_fechas')

# ...y las combinamos con el resumen anterior para calcular el promedio de viajes POR FECHA
# de ese dia de la semana (por ejemplo, viajes totales de todos los "viernes" del mes,
# divididos entre el numero de viernes que hubo).
patron_por_dia = patron_por_dia.merge(fechas_distintas_por_dia, on=['dia_num', 'dia_nombre'])
patron_por_dia['viajes_por_fecha'] = (patron_por_dia['n_viajes'] / patron_por_dia['n_fechas']).round(2)

patron_por_dia
```

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Aplicamos el criterio definido en 2.4: excluir viajes fuera de marzo de 2019, con
# duracion menor a 1 minuto, o con velocidad promedio mayor a 60 mph. Las tres condiciones
# se combinan con OR (|), ya que basta con que se cumpla una de ellas para excluir el viaje.

fuera_de_periodo = ~((taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3))
duracion_menor_1min = taxis['duracion_min'] < 1
# fillna(False): los NaN de velocidad_mph (duracion cero) ya quedan cubiertos por la condicion
# de duracion, asi que no queremos que un NaN en esta comparacion cuente como "True" por error.
velocidad_extrema = (taxis['velocidad_mph'] > 60).fillna(False)

condicion_excluir = fuera_de_periodo | duracion_menor_1min | velocidad_extrema

print(f"Excluidos por estar fuera de marzo de 2019: {fuera_de_periodo.sum()}")
print(f"Excluidos por duracion menor a 1 minuto: {duracion_menor_1min.sum()}")
print(f"Excluidos por velocidad mayor a 60 mph: {velocidad_extrema.sum()}")
print(f"Total de viajes excluidos (union de las tres condiciones): {condicion_excluir.sum()}")

taxis_limpio = taxis[~condicion_excluir].copy()

print(f"\nViajes antes de limpiar: {len(taxis)}")
print(f"Viajes despues de limpiar: {len(taxis_limpio)}")
print(f"Viajes excluidos en total: {len(taxis) - len(taxis_limpio)}")

# Verificamos que taxis_limpio cumple las tres condiciones.
verif_periodo = (~((taxis_limpio['pickup'].dt.year == 2019) & (taxis_limpio['pickup'].dt.month == 3))).sum()
verif_duracion = (taxis_limpio['duracion_min'] < 1).sum()
verif_velocidad = (taxis_limpio['velocidad_mph'] > 60).sum()
print(f"\nVerificacion -> fuera de periodo: {verif_periodo}, duracion<1min: {verif_duracion}, velocidad>60mph: {verif_velocidad}")
print("Todas las verificaciones dan 0" if (verif_periodo, verif_duracion, verif_velocidad) == (0, 0, 0) else "¡Atencion! Quedan casos sin depurar")
```

En conjunto, esto llevaría a un modelo con **mayor error de predicción** y con métricas de evaluación poco representativas del desempeño real sobre viajes típicos, además de que en producción el modelo podría generar predicciones erráticas si estos patrones anómalos se repiten en datos nuevos sin ser detectados.
