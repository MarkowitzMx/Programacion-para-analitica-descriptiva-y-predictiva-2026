## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Contamos las filas duplicadas considerando TODAS las columnas del dataset.
# keep='first' -> marca como duplicada cada copia EXCEPTO la primera aparicion de cada combinacion.
#                 el resultado es el numero de filas "sobrantes" si conservaramos solo la primera copia.
# keep='last'  -> marca como duplicada cada copia EXCEPTO la ultima aparicion.
#                 por eso el conteo es el mismo numero que con 'first' (es la misma cantidad de "sobrantes",
#                 solo cambia CUAL copia se deja como la "original").
# keep=False   -> marca como duplicada CUALQUIER fila que tenga al menos una copia identica en otro lugar,
#                 es decir, TODAS las filas que participan en un grupo repetido (incluida la primera y la ultima).

dup_first = hoteles.duplicated(keep='first').sum()
dup_last = hoteles.duplicated(keep='last').sum()
dup_false = hoteles.duplicated(keep=False).sum()

print(f"keep='first' -> {dup_first} filas duplicadas (filas 'sobrantes' si nos quedamos con la primera copia)")
print(f"keep='last'  -> {dup_last} filas duplicadas (filas 'sobrantes' si nos quedamos con la ultima copia)")
print(f"keep=False   -> {dup_false} filas duplicadas (TODAS las filas involucradas en algun grupo repetido)")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Porcentaje de filas "sobrantes" (las que se eliminarian si depuraramos el dataset)
# respecto al total de filas. Usamos el resultado de keep='first' de la actividad 1.1.
pct_sobrantes = dup_first / len(hoteles) * 100
print(f"Porcentaje de filas sobrantes respecto al total: {pct_sobrantes:.2f}%")

# Numero de grupos repetidos: combinaciones de valores que aparecen mas de una vez.
# Tomamos todas las filas que participan en un duplicado (keep=False) y eliminamos las copias
# dentro de ese subconjunto (drop_duplicates) para quedarnos con UNA fila por cada combinacion distinta.
# El numero de filas resultante es el numero de grupos repetidos.
filas_en_grupos = hoteles[hoteles.duplicated(keep=False)]
n_grupos_repetidos = filas_en_grupos.drop_duplicates().shape[0]
print(f"Numero de grupos repetidos (combinaciones distintas que se repiten): {n_grupos_repetidos}")
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Copiamos el DataFrame para no modificar el original y agregamos una columna booleana
# que indica si la fila es un duplicado, usando keep='first' (asi cada grupo aporta el numero
# de copias "sobrantes" respecto a su primera aparicion).
hoteles_marcado = hoteles.copy()
hoteles_marcado['es_duplicado'] = hoteles_marcado.duplicated(keep='first')

# Agrupamos por hotel y calculamos: cuantas filas sobrantes hay (suma de la columna booleana,
# porque True se comporta como 1) y cuantas reservas totales tiene cada hotel (count).
duplicados_por_hotel = hoteles_marcado.groupby('hotel')['es_duplicado'].agg(
    filas_duplicadas='sum',
    total_reservas='count'
)

# El porcentaje se calcula respecto al total de CADA hotel (no respecto al total general),
# para poder comparar la incidencia del problema entre los dos hoteles de forma justa.
duplicados_por_hotel['porcentaje'] = (
    duplicados_por_hotel['filas_duplicadas'] / duplicados_por_hotel['total_reservas'] * 100
).round(2)

print(duplicados_por_hotel)
```

## Actividad 2. Inspección y criterio de duplicado (25 puntos)

### 2.1 Ver los grupos repetidos (6 pts)

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

`hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adults`, `country`, `market_segment`, `deposit_type`, `adr`, `reservation_status`.

Elige tú las columnas de ordenamiento y explica tu elección en un comentario.

```python
# Seleccionamos todas las filas que participan en algun duplicado (keep=False incluye
# la primera, la ultima y cualquier copia intermedia de cada grupo repetido).
columnas_vista = ['hotel', 'lead_time', 'arrival_date_year', 'arrival_date_month',
                   'arrival_date_day_of_month', 'adults', 'country', 'market_segment',
                   'deposit_type', 'adr', 'reservation_status']

filas_duplicadas = hoteles[hoteles.duplicated(keep=False)]

# Ordenamos por hotel y por la fecha de llegada (anio, mes, dia) y luego por lead_time y adr.
# Elegimos estas columnas porque son las que en teoria deberian coincidir entre copias de una
# MISMA reserva (mismo hotel, misma fecha de llegada). Ordenar por ellas hace que las filas
# identicas queden fisicamente juntas al imprimir la tabla, facilitando la inspeccion visual
# de cada grupo repetido.
filas_duplicadas_ordenadas = filas_duplicadas.sort_values(
    by=['hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 'lead_time', 'adr']
)

filas_duplicadas_ordenadas[columnas_vista].head(12)
```

### 2.2 Tamaño de los grupos (6 pts)

Usa `value_counts()` sobre el `DataFrame` completo para contar cuántas veces aparece cada combinación de filas. Considera que las columnas `agent` y `company` tienen faltantes (revisa el parámetro `dropna`).

Con ese resultado:

1. Muestra **cuántos grupos** hay de cada tamaño (2 copias, 3 copias, …). Basta con los 10 tamaños más frecuentes.
2. Indica el **tamaño del grupo más grande** y muestra su combinación de valores.

```python
# value_counts() sobre el DataFrame completo cuenta cuantas veces se repite cada
# combinacion unica de valores (fila completa). Usamos dropna=False porque las columnas
# 'agent' y 'company' tienen muchos NaN: si no lo indicamos, pandas podria tratar de forma
# especial esas combinaciones con NaN y no contarlas como iguales entre si por defecto en
# versiones antiguas; con dropna=False forzamos a que los NaN se consideren un valor mas
# al comparar, igual que hace duplicated().
conteo_combinaciones = hoteles.value_counts(dropna=False)

# 1. Cuantos grupos hay de cada tamano (2 copias, 3 copias, etc.).
# Tomamos la distribucion de los propios conteos: cuantas combinaciones (grupos) tienen
# tamano 1, cuantas tienen tamano 2, etc. Mostramos los 10 tamanos mas frecuentes.
tamanos_de_grupo = conteo_combinaciones.value_counts().sort_values(ascending=False)
print("Cantidad de grupos segun su tamano (10 mas frecuentes):")
print(tamanos_de_grupo.head(10))

# 2. Tamano del grupo mas grande y su combinacion de valores.
tamano_maximo = conteo_combinaciones.max()
print(f"\nEl grupo mas grande tiene {tamano_maximo} copias identicas.")
print("\nCombinacion de valores de ese grupo:")
print(conteo_combinaciones[conteo_combinaciones == tamano_maximo])
```

### 2.3 ¿En qué se distinguen las filas duplicadas? (6 pts)

Compara las filas involucradas en duplicados (`keep=False`) con el dataset completo en:

- Distribución de `market_segment` (proporciones).
- Distribución de `deposit_type` (proporciones).
- Tasa de cancelación (`is_canceled`).

```python
# Comparamos la distribucion de algunas variables clave entre:
#  a) el dataset completo (referencia)
#  b) el subconjunto de filas involucradas en duplicados (keep=False)

filas_duplicadas = hoteles[hoteles.duplicated(keep=False)]

# Proporciones de market_segment: normalize=True convierte los conteos en proporciones (0 a 1)
# para poder comparar directamente aunque el numero de filas de cada grupo sea distinto.
comparacion_market_segment = pd.concat([
    hoteles['market_segment'].value_counts(normalize=True).rename('dataset_completo'),
    filas_duplicadas['market_segment'].value_counts(normalize=True).rename('duplicados')
], axis=1).round(3)
print("=== Proporcion por market_segment ===")
print(comparacion_market_segment)

# Proporciones de deposit_type
comparacion_deposit_type = pd.concat([
    hoteles['deposit_type'].value_counts(normalize=True).rename('dataset_completo'),
    filas_duplicadas['deposit_type'].value_counts(normalize=True).rename('duplicados')
], axis=1).round(3)
print("\n=== Proporcion por deposit_type ===")
print(comparacion_deposit_type)

# Tasa de cancelacion: como is_canceled es 0/1, la media es directamente la proporcion de
# reservas canceladas.
print("\n=== Tasa de cancelacion (is_canceled) ===")
print(f"Dataset completo: {hoteles['is_canceled'].mean():.4f}")
print(f"Filas duplicadas: {filas_duplicadas['is_canceled'].mean():.4f}")
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Eliminamos los duplicados exactos conservando la primera aparicion de cada combinacion
# (keep='first' es el comportamiento por defecto de drop_duplicates, lo dejamos explicito
# por claridad). reset_index(drop=True) renumera el indice de 0 a n-1 para que quede
# consecutivo despues de haber quitado filas.
filas_antes = len(hoteles)

hoteles_limpio = hoteles.drop_duplicates(keep='first').reset_index(drop=True)

filas_despues = len(hoteles_limpio)
filas_eliminadas = filas_antes - filas_despues

print(f"Filas antes de eliminar duplicados: {filas_antes}")
print(f"Filas despues de eliminar duplicados: {filas_despues}")
print(f"Filas eliminadas: {filas_eliminadas}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Calculamos, para cada version del dataset, tres metricas por hotel:
# tamano del grupo (numero de reservas), tasa de cancelacion promedio y adr promedio.
# Usamos .agg() con un diccionario de nombre->(columna, funcion) para nombrar las columnas
# resultantes directamente y evitar un MultiIndex en las columnas.

resumen_original = hoteles.groupby('hotel').agg(
    n_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    adr_promedio=('adr', 'mean')
).reset_index()

resumen_limpio = hoteles_limpio.groupby('hotel').agg(
    n_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    adr_promedio=('adr', 'mean')
).reset_index()

print("Resumen con datos originales:")
print(resumen_original)
print("\nResumen sin duplicados:")
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Agregamos una columna 'version' a cada resumen para poder distinguir el origen de cada
# fila una vez que los unamos en un solo DataFrame.
resumen_original_v = resumen_original.copy()
resumen_original_v['version'] = 'original'

resumen_limpio_v = resumen_limpio.copy()
resumen_limpio_v['version'] = 'sin duplicados'

# pd.concat() apila los dos resumenes uno debajo del otro (axis=0 por defecto).
# Ordenamos por 'hotel' para que las dos versiones de cada hotel queden juntas y
# renumeramos el indice con reset_index(drop=True) para que quede consecutivo.
comparacion = pd.concat([resumen_original_v, resumen_limpio_v], ignore_index=True)
comparacion = comparacion.sort_values(by='hotel').reset_index(drop=True)

comparacion
```

- **Metricas de evaluacion poco confiables:** el desempeno reportado (exactitud, AUC, etc.) no reflejaria el desempeno real del modelo sobre reservas nuevas, sino un desempeno inflado por la memorizacion de filas repetidas.
